```python
import requests
import wikipedia
import nest_asyncio
from telegram import Update
from telegram.ext import Application, CommandHandler, ContextTypes

nest_asyncio.apply()

wikipedia.set_lang("en")

BOT_TOKEN = "8978660219:AAGhkTL1YRoP2dpvLfnkFQMSu3_Zawm9Tuo"

async def start(update: Update, context: ContextTypes.DEFAULT_TYPE):
    welcome_text = (
        "Hello! I am CityPulse Bot 🏙️\n\n"
        "Available commands:\n"
        "/weather <city> - Get current weather\n"
        "/wiki <topic> - Get a short Wikipedia summary"
    )
    await update.message.reply_text(welcome_text)

async def weather(update: Update, context: ContextTypes.DEFAULT_TYPE):
    if not context.args:
        await update.message.reply_text("Please provide a city name. Example: /weather Lodz")
        return
    
    city = " ".join(context.args)
    url = f"https://wttr.in/{city}?format=j1"
    
    try:
        response = requests.get(url)
        if response.status_code != 200:
            await update.message.reply_text("City not found. Please check the spelling.")
            return
        
        data = response.json()
        temp = data['current_condition'][0]['temp_C']
        desc = data['current_condition'][0]['weatherDesc'][0]['value']
        
        await update.message.reply_text(f"Weather in {city.title()}:\n🌡️ {temp}°C\n☁️ {desc}")
    except Exception:
        await update.message.reply_text("Error fetching weather data. Try again later.")

async def wiki(update: Update, context: ContextTypes.DEFAULT_TYPE):
    if not context.args:
        await update.message.reply_text("Please provide a topic. Example: /wiki Python")
        return
    
    query = " ".join(context.args)
    try:
        summary = wikipedia.summary(query, sentences=2)
        await update.message.reply_text(f"📚 {summary}")
    except wikipedia.exceptions.DisambiguationError:
        await update.message.reply_text("Topic is too broad. Please be more specific.")
    except wikipedia.exceptions.PageError:
        await update.message.reply_text("No Wikipedia page found for this topic.")
    except Exception as e:
        
        await update.message.reply_text("Error fetching Wikipedia data.")
        
        print(f"Wiki error: {e}")
        
def main():
    app = Application.builder().token(BOT_TOKEN).build()
    
    app.add_handler(CommandHandler("start", start))
    app.add_handler(CommandHandler("weather", weather))
    app.add_handler(CommandHandler("wiki", wiki))
    
    print("CityPulse Bot is running (Keyless Version)...")
    app.run_polling()

if __name__ == "__main__":
    main()
```
